```python
import random
import matplotlib.pyplot as plt
```

```python
def two_ball_selection(balls, selection):
    ind_balls = balls/2
    diff_balls = []
    for i in range(selection):
        diff_balls.append(random.randint(0,1))
    diff = []
    for j in range(int(100)):
        a_0, a_1 = list(diff_balls).count(0), list(diff_balls).count(1)
        diff.append(abs(a_0-a_1))
    return sum(diff)/len(diff)

    
```

```python
data = [10,50,100,150,200,250,300,350,400,450,500]
```

```python
dicts = {}
for exp in range(100):
    for k in data:
        if f"{k}" in dicts:
            dicts[f"{k}"] = dicts[f"{k}"].append(two_ball_selection(5000,k))
        elif f"{k}" not in dicts:
            dicts[f"{k}"] = list(two_ball_selection(5000,k))
        
dicts
```
